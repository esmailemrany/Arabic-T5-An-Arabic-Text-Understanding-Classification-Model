```python
import re
import unicodedata

def clean_tags(value):
    # 1) إذا كانت القيمة فارغة
    if pd.isna(value):
        return ""

    # 2) تحويلها إلى نص
    value = str(value)

    # 3) توحيد Unicode
    value = unicodedata.normalize("NFKC", value)

    # 4) إزالة العلامات الخفية
    value = re.sub(r"[\u200b-\u200f\u202a-\u202e\u2060-\u206f\ufeff]", "", value)

    # 5) تقسيم الـ Tags
    tags = value.split("|")

    cleaned = []

    for tag in tags:
        # إزالة المسافات الزائدة
        tag = tag.strip()

        # توحيد المسافات داخل الـ Tag
        tag = re.sub(r"\s+", " ", tag)

        # تجاهل الـ Tag الفارغ
        if not tag:
            continue

        # منع تكرار نفس الـ Tag داخل المقال
        if tag not in cleaned:
            cleaned.append(tag)

    # 6) ترتيب ثابت للـ Tags
    cleaned = sorted(cleaned)

    # 7) إعادة تجميعها
    return " | ".join(cleaned)


df["tags"] = df["tags"].apply(clean_tags)
```

```python
# التصنيفات الرئيسية والفرعية الموجودة فعليًا
valid_main = set(df["main_category"].dropna().unique())
valid_sub = set(df["subcategory"].dropna().unique())

# نفحص كل Target
def check_target(row):
    parts = [x.strip() for x in row["target"].split("|")]

    # لازم يكون عندنا على الأقل:
    # main_category + subcategory
    if len(parts) < 2:
        return False

    main = parts[0]
    sub = parts[1]

    # التأكد أن الرئيسي والفرعي صحيحان
    if main not in valid_main:
        return False

    if sub not in valid_sub:
        return False

    # منع القيم الفارغة
    if main == "" or sub == "":
        return False

    return True


df["target_valid"] = df.apply(check_target, axis=1)

print("عدد Targets الصحيحة:", df["target_valid"].sum())
print("عدد Targets الخاطئة:", (~df["target_valid"]).sum())
```

```python
import re
import html

def clean_title(text):
    # 1) تحويل HTML entities مثل &quot; و &amp;
    text = html.unescape(str(text))

    # 2) إزالة المحارف غير المرئية/التالفة
    text = re.sub(r"[\u200b-\u200f\u202a-\u202e\ufeff]", "", text)

    # 3) إزالة محارف التحكم غير الضرورية
    text = re.sub(r"[\x00-\x08\x0b\x0c\x0e-\x1f\x7f]", " ", text)

    # 4) توحيد المسافات
    text = re.sub(r"\s+", " ", text)

    # 5) إزالة المسافات في البداية والنهاية
    text = text.strip()

    return text
```

```python
import re

def find_weird_chars(text):
    """
    تبحث عن المحارف غير المرغوبة أو المشوهة داخل العنوان.
    """
    if not isinstance(text, str):
        return []

    weird = []

    for char in text:
        code = ord(char)

        # محارف التحكم
        if code < 32 and char not in "\n\t":
            weird.append(char)

        # محارف Unicode الاتجاهية/غير المرئية
        elif 0x200B <= code <= 0x200F:
            weird.append(char)

        elif 0x202A <= code <= 0x202E:
            weird.append(char)

        # محرف الاستبدال الذي يدل غالبًا على تلف في الترميز
        elif char == "�":
            weird.append(char)

    return weird


# تطبيق الفحص
weird_chars = df["clean_title"].apply(find_weird_chars)

# العناوين التي تحتوي على محارف غريبة
weird_rows = df[weird_chars.apply(len) > 0][
    ["title", "clean_title"]
]

print("عدد العناوين التي تحتوي على محارف غريبة:", len(weird_rows))

weird_rows.head(50)
```

```python
from datasets import Dataset

# ==========================================
# بناء الـTarget
# ==========================================

def build_target(target):
    parts = [part.strip() for part in str(target).split("|")]

    main_category = parts[0]
    sub_category = parts[1]
    tags = " | ".join(parts[2:])

    return (
        f"<MAIN> {main_category} "
        f"<SUB> {sub_category} "
        f"<TAGS> {tags}"
    )


# ==========================================
# تحويل كل DataFrame إلى Dataset
# ==========================================

def tokenize_dataset(df):

    dataset = Dataset.from_pandas(
        df[["title", "target"]],
        preserve_index=False
    )

    def tokenize_function(examples):

        inputs = examples["title"]

        targets = [
            build_target(target)
            for target in examples["target"]
        ]

        model_inputs = tokenizer(
            inputs,
            max_length=64,
            truncation=True,
            padding="max_length"
        )

        labels = tokenizer(
            targets,
            max_length=64,
            truncation=True,
            padding="max_length"
        )

        model_inputs["labels"] = labels["input_ids"]

        return model_inputs

    tokenized = dataset.map(
        tokenize_function,
        batched=True,
        remove_columns=dataset.column_names
    )

    return tokenized


# ==========================================
# Tokenization
# ==========================================

train_tokenized = tokenize_dataset(train_df)

val_tokenized = tokenize_dataset(val_df)

test_tokenized = tokenize_dataset(test_df)


# ==========================================
# النتائج
# ==========================================

print("=" * 60)
print("TOKENIZATION COMPLETE")
print("=" * 60)

print("TRAIN      :", len(train_tokenized))
print("VALIDATION :", len(val_tokenized))
print("TEST       :", len(test_tokenized))

print("\nCOLUMNS:")
print(train_tokenized.column_names)
```
